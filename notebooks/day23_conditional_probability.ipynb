Understanding Bayes' rule
Our Day 23 roadmap asks us to understand conditional probability and apply it to a rare-event detection problem.

Bayes' rule is:
\[
P(I\mid A)=\frac{P(A\mid I)\times P(I)}{P(A)}
\]

Where:
- \(I\) = Event is an actual incident.
- \(A\) = Detector generates an alert.
- \(P(A\mid I)\) = Sensitivity, or 0.90.
- \(P(I)\) = Incident base rate, or 0.01.
- \(P(A)\) = Probability of generating an alert.

```python
total_events = 10000
incident_rate = 0.01
sensitivity = 0.90
false_positive_rate = 0.05
```

```python
actual_incidents = total_events * incident_rate
non_incidents = total_events - actual_incidents
true_alerts = actual_incidents * sensitivity
false_alerts = non_incidents * false_positive_rate

print("Actual Incidents: ", actual_incidents)
print("Non Incidents : ", non_incidents)
print("True Alerts: ", true_alerts)
print("False Alerts: ", false_alerts)
```

```python
probability_of_alert = (
    (sensitivity * incident_rate)
    + false_positive_rate * (1 - incident_rate)
)

print(f"Probability of Alert:" , round((probability_of_alert * 100), 2))
```

```python
posterior_probability = (sensitivity * incident_rate) / probability_of_alert

print("Posterior Probability (Precision) :", round((posterior_probability * 100), 2))
```

```python
# Testing what happens to Probabilty of Alert and Precision when incident rate changes from 1% to 10%

for incident_rate in [0.01, 0.10]:
    p_alert = (
        sensitivity * incident_rate
        + false_positive_rate * (1 - incident_rate)
    )

    precision = (sensitivity * incident_rate) / p_alert

    print(
        f"Incident rate: {incident_rate:.0%} | "
        f"Alert probability: {p_alert:.2%} | "
        f"Precision: {precision:.2%}"
    )
```
